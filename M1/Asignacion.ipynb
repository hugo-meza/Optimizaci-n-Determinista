```python
import pulp

# 1. Inicializar el modelo definiendo que es un problema de MINIMIZACIÓN
prob = pulp.LpProblem("Problema_de_Asignacion", pulp.LpMinimize)

# 2. Definir nuestras listas (nodos)
asignados = ['A', 'B', 'C', 'D']
tareas = ['1', '2', '3', '4']

# 3. Crear un diccionario con la matriz de costos original
costos = {
    'A': {'1': 8, '2': 6, '3': 5, '4': 7},
    'B': {'1': 6, '2': 5, '3': 3, '4': 4},
    'C': {'1': 7, '2': 8, '3': 4, '4': 6},
    'D': {'1': 6, '2': 7, '3': 5, '4': 6}
}

# 4. Crear las variables de decisión
# Serán variables binarias (0 o 1). 1 significa que la asignación se realiza.
x = pulp.LpVariable.dicts("Asignacion", (asignados, tareas), cat='Binary')

# 5. Función Objetivo
# Queremos minimizar la suma de los costos de las asignaciones que se hagan
prob += pulp.lpSum([costos[i][j] * x[i][j] for i in asignados for j in tareas]), "Costo_Total"

# 6. Restricciones
# a) Cada 'Asignado' debe tener exactamente una 'Tarea'
for i in asignados:
    prob += pulp.lpSum([x[i][j] for j in tareas]) == 1, f"Asignado_{i}_una_sola_tarea"

# b) Cada 'Tarea' debe ser realizada por exactamente un 'Asignado'
for j in tareas:
    prob += pulp.lpSum([x[i][j] for i in asignados]) == 1, f"Tarea_{j}_un_solo_asignado"

# 7. Resolver el modelo
prob.solve()

# 8. Mostrar los resultados
print(f"Estado de la resolución: {pulp.LpStatus[prob.status]}\n")
print("--- Asignaciones Óptimas ---")

for i in asignados:
    for j in tareas:
        # Si el valor de la variable es 1 (o muy cercano por redondeo interno), se asignó
        if pulp.value(x[i][j]) == 1.0:
            print(f"{i} realiza la Tarea {j}  --> Costo: ${costos[i][j]}")

print("-" * 28)
print(f"Costo Mínimo Total: ${int(pulp.value(prob.objective))}")
```

```python
# 1. Inicializar el modelo definiendo que es un problema de MINIMIZACIÓN
prob = pulp.LpProblem("Natacion", pulp.LpMinimize)

# 2. Definir nuestras listas (nodos)
asignados = ['Carl', 'Chris', 'David', 'Tony', 'Ken']
tareas = ['Dorso', 'Pecho', 'Mariposa', 'Libre','Nada']

# 3. Crear un diccionario con la matriz de costos original
costos = {
    'Carl': {'Dorso': 37.7, 'Pecho': 43.4, 'Mariposa': 33.3, 'Libre': 29.2, 'Nada': 99},
    'Chris': {'Dorso': 32.9, 'Pecho': 33.1, 'Mariposa': 28.5, 'Libre': 26.4, 'Nada': 99},
    'David': {'Dorso': 33.8, 'Pecho': 42.2, 'Mariposa': 38.9, 'Libre': 29.6, 'Nada': 99},
    'Tony': {'Dorso': 37.0, 'Pecho': 34.7, 'Mariposa': 30.4, 'Libre': 28.5, 'Nada': 99},
    'Ken': {'Dorso': 35.4, 'Pecho': 41.8, 'Mariposa': 33.6, 'Libre': 31.1, 'Nada': 99}
}

# 4. Crear las variables de decisión
# Serán variables binarias (0 o 1). 1 significa que la asignación se realiza.
x = pulp.LpVariable.dicts("Asignacion", (asignados, tareas), cat='Binary')

# 5. Función Objetivo
# Queremos minimizar la suma de los costos de las asignaciones que se hagan
prob += pulp.lpSum([costos[i][j] * x[i][j] for i in asignados for j in tareas]), "Costo_Total"

# 6. Restricciones
# a) Cada 'Asignado' debe tener exactamente una 'Tarea'
for i in asignados:
    prob += pulp.lpSum([x[i][j] for j in tareas]) == 1, f"Asignado_{i}_una_sola_tarea"

# b) Cada 'Tarea' debe ser realizada por exactamente un 'Asignado'
for j in tareas:
    prob += pulp.lpSum([x[i][j] for i in asignados]) == 1, f"Tarea_{j}_un_solo_asignado"

# 7. Resolver el modelo
prob.solve()

# 8. Mostrar los resultados
print(f"Estado de la resolución: {pulp.LpStatus[prob.status]}\n")
print("--- Asignaciones Óptimas ---")

for i in asignados:
    for j in tareas:
        # Si el valor de la variable es 1 (o muy cercano por redondeo interno), se asignó
        if pulp.value(x[i][j]) == 1.0:
            if j != "Nada":
                print(f"{i} realiza la Tarea {j}  --> Tiempo: {costos[i][j]} s")
            else:
                ai = i
                aj = j

print("-" * 28)
print(f"Tiempo Mínimo Total: {int(pulp.value(prob.objective) - costos[ai][aj])} s")
```
