## Ejemplo 1

```python
'''
Ejemplo 1
x1: cantidad de automóviles a fabricar
x2: cantidad de vagonetas a fabricar
'''
model = pulp.LpProblem("Maximize_Profit", pulp.LpMaximize)

x1 = pulp.LpVariable('x1', lowBound=0, cat='Integer')
x2 = pulp.LpVariable('x2', lowBound=0, cat='Integer')

model += 3000 * x1 + 4000 * x2, "Profit"

model += x1 <= 300, "Constraint1"
model += x2 <= 200, "Constraint2"
model += 2 * x1 + 3 * x2 <= 900, "Constraint3"

model.solve()
print(f"Status: {pulp.LpStatus[model.status]}")
print(f"Result: {pulp.value(model.objective)}")
      
print(f"Cantidad de automóviles a fabricar: {pulp.value(x1)}")
print(f"Cantidad de vagonetas a fabricar: {pulp.value(x2)}")
```

```python
for constraint in model.constraints:
    constraint_sum = 0
    for var, coefficient in model.constraints[constraint].items():
        constraint_sum += var.varValue * coefficient
    print(model.constraints[constraint].name, constraint_sum)
```

## Ejemplo 7

Un inversionista dispone de $10 000 para invertirlos en cuatro proyectos. La tabla siguiente presenta el flujo de efectivo para las cuatro inversiones:


| Proyecto | Año 1 | Año 2 | Año 3 | Año 4 | Año 5 |
| :---: | :---: | :---: | :---: | :---: | :---: |
| 1 | -1 | 0.5 | 0.3 | 1.8 | 1.2 |
| 2 | -1 | 0.6 | 0.2 | 1.5 | 1.3 |
| 3 | 0 | -1 | 0.8 | 1.9 | 0.8 |
| 4 | -1 | 0.4 | 0.6 | 1.8 | 0.95 |


La información que aparece en la tabla puede interpretarse como sigue: para el proyecto 1, 1 invertido al inicio del año 1 redituará (rendimiento) 0.5 al inicio del año 2; 0.3 al inicio del año 3; 1.8 al inicio del año 4 y 1.2 al inicio del año 5. El inversionista tiene opción adicional de invertir en
una cuenta bancaria que gana 6.5% anual. Todos los fondos acumulados al final del año 1 pueden volverse a invertir en el año siguiente. Formule el problema como un programa lineal para determinar la asignación óptima de fondos a oportunidades de inversión.


$I = \{1, 2, 3, 4\}$,   $J = \{1, 2, 3, 4, 5\}$


$x_i: \text{Cantidad invertida en el proyecto i, } i \in I$

$y_j: \text{Cantidad invertida en la cuenta bancaria en el anio j, } j \in I$

$\text{max } z = y_5$

S.A:

$\quad x_1 + x_2 + x_4 + y_1 = 10000$

$\quad 0.5 x_1 + 0.6 x_2 + 0.4 x_4 +1.065 y_1 = x_3 + y_2$

$\quad 0.3 x_1 + 0.2 x_2 + 0.8 x_3 + 0.6 x_4 + 1.065 y_2 = y_3$

$\quad 1.8 x_1 + 1.5 x_2 + 1.9 x_3 + 1.8 x_4 + 1.065 y_3 = y_4$

$\quad 1.2 x_1 + 1.3 x_2 + 0.8 x_3 + 0.95 x_4 + 1.065 y_4 = y_5$

$\quad x_i,  y_j \geq 0$

```python
model7 = pulp.LpProblem("Maximize_Profit", pulp.LpMaximize)

# xi: Cantidad invertida en el proyecto i
x1 = pulp.LpVariable('Proyecto_1', lowBound=0, cat='Continuous')
x2 = pulp.LpVariable('Proyecto_2', lowBound=0, cat='Continuous')
x3 = pulp.LpVariable('Proyecto_3', lowBound=0, cat='Continuous')
x4 = pulp.LpVariable('Proyecto_4', lowBound=0, cat='Continuous')

# yi: Cantidad invertida en la cuenta bancaria en el año j
y1 = pulp.LpVariable('Year_1', lowBound=0, cat='Continuous')
y2 = pulp.LpVariable('Year_2', lowBound=0, cat='Continuous')
y3 = pulp.LpVariable('Year_3', lowBound=0, cat='Continuous')
y4 = pulp.LpVariable('Year_4', lowBound=0, cat='Continuous')
y5 = pulp.LpVariable('Year_5', lowBound=0, cat='Continuous')

model7 += y5, "Profit"

model7 += x1  + x2 + x4 + y1 == 10000, ""
model7 += x2 <= 200, "Constraint2"
model7 += 2 * x1 + 3 * x2 <= 900, "Constraint3"

model7.solve()

print(f"Estado de la solución: {pulp.LpStatus[model7.status]}\n")

print("--- Estrategia de Inversión Óptima ---")
print(f"Proyecto 1 (x1): ${x1.varValue:.2f}")
print(f"Proyecto 2 (x2): ${x2.varValue:.2f}")
print(f"Proyecto 3 (x3): ${x3.varValue}")
print(f"Proyecto 4 (x4): ${x4.varValue:.2f}")

print("\n--- Estrategia de Ahorro Bancario ---")
print(f"Banco Año 1 (y1): ${y1.varValue:.2f}")
print(f"Banco Año 2 (y2): ${y2.varValue}")
print(f"Banco Año 3 (y3): ${y3.varValue}")
print(f"Banco Año 4 (y4): ${y4.varValue}")

# Capital final
capital_final = pulp.value(model7.objective)
print(f"\nCapital máximo acumulado para el Año 5: ${capital_final:,.2f}")
```
