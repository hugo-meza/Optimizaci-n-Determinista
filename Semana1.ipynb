## Ejemplo 1

```python
'''
Ejemplo 1
x1: cantidad de automóviles a fabricar
x2: cantidad de vagonetas a fabricar
'''
model = pulp.LpProblem("Maximize_Profit", pulp.LpMaximize)
```

```python
x1 = pulp.LpVariable('x1', lowBound=0, cat='Integer')
x2 = pulp.LpVariable('x2', lowBound=0, cat='Integer')
```

```python
model += 3000 * x1 + 4000 * x2, "Profit"
```

```python
model += x1 <= 300, "Constraint1"
model += x2 <= 200, "Constraint2"
model += 2 * x1 + 3 * x2 <= 900, "Constraint3"
```

```python
model.solve()
print(f"Status: {pulp.LpStatus[model.status]}")
print(f"Result: {pulp.value(model.objective)}")
```

```python
print(f"Cantidad de automóviles a fabricar: {pulp.value(x1)}")
print(f"Cantidad de vagonetas a fabricar: {pulp.value(x2)}")
```

```python
# iterate over the variables of the constraint and sum their values
# I'm not considering here the possibility that varValue may be None
for constraint in model.constraints:
    constraint_sum = 0
    for var, coefficient in model.constraints[constraint].items():
        constraint_sum += var.varValue * coefficient
    print(model.constraints[constraint].name, constraint_sum)
```
